```python
class Enfermero:
    def __init__(self, nombre, telefono, edad):
        # Atributos de la clase
        self.nombre = nombre
        self.telefono = telefono
        self.edad = edad

    # Función / Método: Educar
    def educar(self):
        return f"El enfermero {self.nombre} está impartiendo una charla sobre prevención de enfermedades."

    # Función / Método: Apoyar
    def apoyar(self):
        return f"El enfermero {self.nombre} está brindando soporte emocional al paciente en recuperación."

    # Función / Método: Atender
    def atender(self):
        return f"El enfermero {self.nombre} está revisando los signos vitales y administrando medicamentos."

# --- Creación de Objetos (Instancias) ---

# Objeto 1
enfermero1 = Enfermero("Carlos Ruiz", "555-0123", 29)

# Objeto 2
enfermero2 = Enfermero("Elena Gómez", "555-9876", 34)

# --- Ejecución y Mensajes ---

print("--- Datos del Enfermero 1 ---")
print(f"Nombre: {enfermero1.nombre} | Tel: {enfermero1.telefono} | Edad: {enfermero1.edad}")
print(enfermero1.educar())
print(enfermero1.atender())

print("\n--- Datos del Enfermero 2 ---")
print(f"Nombre: {enfermero2.nombre} | Tel: {enfermero2.telefono} | Edad: {enfermero2.edad}")
print(enfermero2.apoyar())
print(enfermero2.atender())

print("Programa terminado")
print("Programa realizado por Johan Paulo Nl = 04")
```
